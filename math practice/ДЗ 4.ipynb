5. Стандартизуйте дані:

```python
def data_scale(data, scaler_type='std'):
    from sklearn.preprocessing import MinMaxScaler
    from sklearn.preprocessing import StandardScaler
    from sklearn.preprocessing import Normalizer
    if scaler_type == 'minmax':
        scaler = MinMaxScaler()
    if scaler_type == 'std':
        scaler = StandardScaler()
    if scaler_type == 'norm':
        scaler = Normalizer()

    scaler.fit(data)
    res = scaler.transform(data)
    return res
```

```python
import pandas as pd


X_scaled = data_scale(df)
df_scaled = pd.DataFrame(X_scaled, columns=df.columns)
df_scaled.head(10)
```

```python
from scipy.spatial.distance import euclidean, cityblock, minkowski, jaccard, hamming, squareform, cosine, chebyshev
import numpy as np


def compute_distance_matrix(X, n_samples, metric="euclidean", p=4):
    # Ініціалізуємо матрицю подібностей
    D = np.zeros((n_samples, n_samples))
    for i in range(n_samples):
        for j in range(i, n_samples):
            match metric:
                case "euclidean":
                    dist = euclidean(X[i], X[j])
                case "manhattan":
                    dist = cityblock(X[i], X[j])
                case "minkowski":
                    dist = minkowski(X[i], X[j], p=p)
                case "jaccard":
                    dist = jaccard(X[i] > 0, X[j] > 0)
                case "hamming":
                    dist = hamming(X[i] > 0, X[j] > 0)
                case "cosine":
                    dist = cosine(X[i], X[j])
                case "chebyshev":
                    dist = chebyshev(X[i], X[j])
                case _:
                    raise ValueError("Unknown metric")
            # Додаємо значення метрик до матриці подібностей
            D[i, j] = dist
            D[j, i] = dist
    return D

# Обчислимо кількість зразків та ознак
n_samples, n_features = X.shape # X = data.data.shape

similarity_matrix = compute_distance_matrix(X_scaled, n_samples)

print("Матриця подібностей:")
print(similarity_matrix)
print(similarity_matrix.shape)
```

```python
from sklearn.metrics import pairwise_distances


def plot_distance_heatmap(data, metric):
    dist = pairwise_distances(data, metric=metric)
    plt.figure(figsize=(6, 5))
    sns.heatmap(dist, cmap='mako')
    plt.title(f'Теплова карта: {metric}')
    plt.xlabel('Зразки')
    plt.ylabel('Зразки')
    plt.tight_layout()
    plt.show()

metrics = ["euclidean", "cosine", "cityblock"]
for metric in metrics:
    plot_distance_heatmap(df_scaled, metric)
```

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import pairwise_distances, adjusted_rand_score


true_labels = data.target

ari_scores = {}

print("🔎 ARI для кожної метрики:")
for metric in metrics:
    dist = pairwise_distances(df, metric=metric)
    model = AgglomerativeClustering(n_clusters=2, metric='precomputed', linkage='average')
    pred_labels = model.fit_predict(dist)

    ari = adjusted_rand_score(true_labels, pred_labels)
    ari_scores[metric] = ari
    print(f"{metric:<10} ARI = {ari:.3f}")

plt.figure(figsize=(8, 5))
sns.barplot(x=list(ari_scores.keys()), y=list(ari_scores.values()), palette='viridis')
plt.title("ARI (Adjusted Rand Index) для різних метрик")
plt.xlabel("Метрика відстані")
plt.ylabel("ARI")
plt.ylim(0, 1.05)
plt.tight_layout()
plt.show()
```
